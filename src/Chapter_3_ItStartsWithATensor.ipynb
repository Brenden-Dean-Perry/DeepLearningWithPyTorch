### Tensor initialization:

```python
# Initialize a tensor of zeros to get appropriate shape
points = torch.zeros(6)

# Set specific elements
points[0] = 4.0
points[1] = 1.0
points[2] = 5.0
points[3] = 3.0
points[4] = 2.0
points[5] = 1.0
```

```python
points = torch.tensor([4.0, 1.0, 5.0, 3.0, 2.0, 1.0])
points
```

```python
float(points[0]), float(points[1])
```

```python
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points
```

```python
points.shape
```

```python
points = torch.zeros(3, 2)
points
```

```python
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points
```

```python
points[0, 1]
```

```python
points[0]
```

The output is another tensor that presents a different view of the same underlying
data. The new tensor is a 1D tensor of size 2, referencing the values of the first row in
the points tensor. Does this mean a new chunk of memory was allocated, values were
copied into it, and the new memory was returned wrapped in a new tensor object? No,
because that would be very inefficient, especially if we had millions of points.

### Indexing tensors

```python
some_list = list(range(6))
some_list[:]
```

```python
some_list[1:4]
```

```python
some_list[1:]
```

```python
some_list[:4]
```

```python
some_list[:-1] # From the start of the list to one before the last element
```

```python
some_list[1:4:2] # Index 1 to 4, exclusive. Step by 2
```

```python
points
```

```python
points[1:] # Implicit all columns
```

```python
points[1:, :] # Explicit all columns
```

```python
points[1:, 0] # All rows after the first; first (zero-index) column
```

```python
points[None] # Adds a dimension of size 1; the same can be accomplished with points.unsqueeze(dim=0)
```

In addition to using ranges, PyTorch features a powerful form of indexing, called
advanced indexing, which we will look at in the next chapter.
